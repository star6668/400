```python
import pandas as pd

from graph_lib import *
```

```python
TEST_CASES = [
    {'n': 100, 'metric': 'clique', 'target': 8},
    {'n': 100, 'metric': 'degeneracy', 'target': 20},
    {'n': 200, 'metric': 'clique', 'target': 15},
    {'n': 200, 'metric': 'degeneracy', 'target': 30},
    {'n': 500, 'metric': 'clique', 'target': 25},
    {'n': 500, 'metric': 'degeneracy', 'target': 60},
    {'n': 1000, 'metric': 'clique', 'target': 17},
    {'n': 1000, 'metric': 'degeneracy', 'target': 31},
    {'n': 1500, 'metric': 'clique', 'target': 20},
    {'n': 1500, 'metric': 'degeneracy', 'target': 45},
]
```

```python
rows = []

for i, case in enumerate(TEST_CASES, start=1):
    n = case['n']
    metric = case['metric']
    target = case['target']

    p, predicted = choose_p(target, metric, n)

    if metric == 'clique':
        G = get_dag(n, clique_size=target, seed=1000 + i)
    else:
        G = get_dag(n, k_degen=target, seed=1000 + i)

    measured = measure_cocomp_property(G, metric)

    rows.append({
        'n': n,
        'metric': metric,
        'target': target,
        'p': round(p, 6),
        'measured': measured,
    })

results = pd.DataFrame(rows)
results
```
